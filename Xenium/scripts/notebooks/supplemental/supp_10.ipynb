```python



sc.settings._vector_friendly = True
os.makedirs(fig_outdir, exist_ok=True)

ax = sc.pl.umap(all_cells_adata, color="EPCAM", vmin="p5", vmax="p95", size=0.5, cmap="Reds", show=False)
ax.set_axis_off()
plt.savefig(os.path.join(fig_outdir, f"all_cells_umap_epcam.pdf"), dpi=300, bbox_inches="tight")
plt.show()

ax = sc.pl.umap(all_cells_adata, color=res, show=False, size=0.5, title="", palette=colors_dutch_y)
ax.set_axis_off()
plt.savefig(os.path.join(fig_outdir, f"all_cells_umap.pdf"), dpi=300, bbox_inches="tight")
plt.show()

sc.tl.rank_genes_groups(all_cells_adata, res, layer='normalized', use_raw=False, dendrogram=False)
sc.pl.rank_genes_groups_dotplot(all_cells_adata, n_genes=5, groupby=res,
                            title=f"", standard_scale="var", show=False, dendrogram=False)
plt.savefig(os.path.join(fig_outdir, f"all_cells_dotplot.pdf"), dpi=300, bbox_inches="tight")
plt.show()

sc.settings._vector_friendly = False
```

```python
all_cells_adata = all_cells_adata[~all_cells_adata.obs['combo_all_cells_0.6'].isin(["10"])].copy()
```

```python
epi_clusters = ["1", "4", "6", "9", "11", "0"]
all_cells_adata.obs["broad_cell_type"] = (
    all_cells_adata.obs["combo_all_cells_0.6"]
    .astype(str)
    .isin(epi_clusters)
    .map({True: "Epithelial", False: "Nonepithelial"})
    .astype("category")
)
```

```python
sc.settings._vector_friendly = True

ax = sc.pl.umap(all_cells_adata, color="broad_cell_type", show=False, size=0.5,
                title="", palette={"Epithelial": "#009432", "Nonepithelial": "#5758BB"})
ax.set_axis_off()
plt.savefig(os.path.join(fig_outdir, "all_cells_umap_broad_celltype.pdf"), dpi=300, bbox_inches="tight")
plt.show()

sc.settings._vector_friendly = False
```
